```python
import requests
```

```python
ticker = "AAPL"
multiplier = "5"
timespan = "minute"
from_date = "2026-10-16"
to_date = "2026-10-16"
path = f"/v2/aggs/ticker/{ticker}/range/{multiplier}/{timespan}/{from_date}/{to_date}"
url = f"https://api.polygon.io{path}?apiKey={POLYGON_API_KEY}"
```

```python
response = requests.get(url)
```
